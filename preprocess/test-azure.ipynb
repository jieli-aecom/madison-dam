```python
%reload_ext autoreload
%autoreload 2

import os
import sys

workspace = os.getcwd()
sys.path.append(f"{workspace}/preprocess")

from azure.storage.blob import BlobServiceClient
from config import AZURE_STORAGE_CONNECTION_STRING, CONTAINER_NAME
import json
```

```python
BLOB_NAME = "actuals.json"

blob_service_client = BlobServiceClient.from_connection_string(AZURE_STORAGE_CONNECTION_STRING)
blob_client = blob_service_client.get_blob_client(container=CONTAINER_NAME, blob=BLOB_NAME)
```

```python
# READ: Download and parse JSON from blob
try:
    blob_data = blob_client.download_blob().readall()
    data = json.loads(blob_data)
    print(f"Read {len(data)} entries from blob.")
except Exception as e:
    data = []
    print(f"No existing blob found or error reading: {e}")
```

```python
# WRITE: Upload JSON to blob (overwrite)
json_data = json.dumps(data, indent=4)
blob_client.upload_blob(json_data, overwrite=True)
print(f"Wrote {len(data)} entries to blob.")
```
